union() method returns a new set with all items from both sets.
update() method inserts the items in set2 into set1. It does not return a new set, but updates set1 with the items from set2.
intersection() method returns a new set with items that are common to both sets.
intersection_update() method removes the items in this set that are not present in other, specified set
difference() method returns a new set with all the items from the first set that are not in the second set.
difference_update() method removes the items in this set that are also present in other, specified set
symmetric_difference() method returns a new set with items in either set, but not in both sets.
symmetric_difference_update() method updates a set with the symmetric difference of itself and another.

Python RegEX


\d → digit (0–9)

\w → word character (a–z, A–Z, 0–9, _)

. → যেকোনো একক ক্যারেক্টার

+ → এক বা একাধিক

* → শূন্য বা একাধিক

? → শূন্য বা একবার

{n} → ঠিক n বার

^ → লাইনের শুরু

$ → লাইনের শেষ

[abc] → a, b, বা c এর যেকোনো একটি

[a-z] → ছোট হাতের অক্ষর a থেকে z

(pattern) → গ্রুপ বানানো


বেসিক উদাহরণ
1. সংখ্যা খুঁজে বের করা
regex
\d+
\d মানে digit (0–9)

+ মানে এক বা একাধিক
👉 টেক্সটে যত সংখ্যা আছে সব বের করবে।

2. নির্দিষ্ট দৈর্ঘ্যের সংখ্যা
regex
\d{4}
{4} মানে ঠিক ৪টা digit
👉 যেমন "2026" বা "1234" ম্যাচ করবে।

3. শব্দ খুঁজে বের করা
regex
\w+
\w মানে word character (a–z, A–Z, 0–9, _)

+ মানে এক বা একাধিক
👉 টেক্সটের সব শব্দ আলাদা করবে।

4. ইমেইল ঠিকানা
regex
[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-z]{2,}
👉 user@example.com এর মতো ইমেইল ম্যাচ করবে।

5. বাংলাদেশি মোবাইল নাম্বার
regex
01[3-9]\d{8}
01 দিয়ে শুরু

[3-9] মানে দ্বিতীয় ডিজিট ৩–৯ এর মধ্যে

\d{8} মানে বাকি ৮টা digit
👉 যেমন 01712345678 ম্যাচ করবে।

6. তারিখ (DD/MM/YYYY)
``regex
\d{2}/\d{2}/\d{4}
```
👉 26/02/2026 এর মতো তারিখ ম্যাচ করবে।

````python
import re

text = "আমার ইমেইল হলো 15we234234326"
pattern = r'\d[0-9]+\d[0-9]'

match = re.search(pattern, text)
if match:
    print("ইমেইল পাওয়া গেছে:", match.group())
````
